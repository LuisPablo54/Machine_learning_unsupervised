## Definimos nuestro modelo

```python
'''
Modelo tradicional, pero también el menos común en entonos de trabajo

model = nn.Sequential(
    nn.Linear(8, 12),
    nn.ReLU(),
    nn.Linear(12, 8),
    nn.ReLU(),
    nn.Linear(8, 1),
    nn.Sigmoid()
)
print(model)

'''
```

```python
# En forma de Clase
class PrimaCalssifier(nn.Module):
    def __init__(self):
        super().__init__()
        self.hidden1 = nn.Linear(8, 12)
        self.act1 = nn.ReLU()
        self.hidden2 = nn.Linear(12, 8)
        self.act2 = nn.ReLU()
        self.output = nn.Linear(8, 1)
        self.act_output = nn.Sigmoid

    def forward(self, x):
        x = self.act1(self.hidden1, (x))
        x = self.act2(self.hidden2, (x))
        x = self.act_output(self.output, (x))
        return(x)
    

model = PrimaCalssifier()
        
```

```python
print(model)
```

## Pereparación para el entrenamiento

```python
loss_fn = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
```
