```python
# Exmaple of gradient descen
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Inicialt parameters
num_iteraciones = 20
lr = 0.1 # Learning rate
w_0 = -2 # Initial point
x = np.linspace(-5, 5, 100)
y = x**2 # Objective function
```

```python
# Objective function
def f(x):
    return x**2 

# Derivative of the objective function
def df(x):
    return 2*x
```

```python
# Start gradient descent
w = w_0
losses = []
positions = []
for i in range(num_iteraciones):
    w = w - lr*df(w)
    positions.append(w)
    loss = f(w)
    losses.append(loss)
```

```python
# Figure 1: Gradient descent
plt.figure(figsize=(5, 4))
plt.plot(x, y, label='$f(x)$', linewidth=2.5, color='#2c3e50')
plt.plot(w, f(w), 'o', markersize=10, label='Initial point', color='#e74c3c')
plt.plot(positions, f(np.array(positions)), 'o-', markersize=6, label='Positions', color='#3498db', alpha=0.7, markevery=1)
plt.legend(fontsize=12, loc='best')
plt.xlabel('$x$', fontsize=14)
plt.ylabel('$f(x)$', fontsize=14)
plt.title('Gradient descent - Evolution of the algorithm', fontsize=16, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# Figure 2: Evolution of the loss
plt.figure(figsize=(5, 4))
plt.plot(loss, linewidth=2.5, color='#e67e22', marker='o', markersize=4)
plt.xlabel('Iterations', fontsize=14)
plt.ylabel('Loss', fontsize=14)
plt.title('Evolution of the loss during the training', fontsize=16, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.gca().xaxis.set_major_formatter(plt.FormatStrFormatter('%d'))
plt.tight_layout()
plt.show()
```
