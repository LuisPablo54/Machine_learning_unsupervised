# Redes Neuronales Convolucionales (CNN / ConvNet)

## Introducción
Las Redes Neuronales Convolucionales son una clase de modelos de deep learning especializados en el procesamiento de datos con una topología en forma de rejilla (grid-like), siendo su aplicación más exitosa el análisis de imágenes y videos. A diferencia de las redes completamente conectadas (Fully Connected), las CNN explotan jerarquías espaciales locales mediante operaciones de convolución, lo que reduce drásticamente la cantidad de parámetros y permite la extracción automática de características relevantes (bordes, texturas, formas, objetos completos).

## Terminos

| Término | Explicación breve |
|---------|-------------------|
| **Convolución** | Operación matemática que aplica un filtro (kernel) sobre la entrada para detectar patrones locales. |
| **Kernel / Filtro** | Matriz pequeña de pesos aprendibles que se desliza sobre la entrada para generar un mapa de activaciones. |
| **Feature Map** | Salida de una capa convolucional; representa la presencia de características específicas en diferentes posiciones espaciales. |
| **Stride (Paso)** | Número de píxeles que se desplaza el kernel en cada movimiento. Stride > 1 reduce la resolución espacial. |
| **Padding (Relleno)** | Adición de ceros alrededor del borde de la entrada para controlar el tamaño de salida y preservar información de bordes. |
| **Capa de Pooling** | Operación de submuestreo (max pooling o average pooling) que reduce dimensiones espaciales y se queda con la información más relevante. |
| **Canal (Channel)** | Cada profundidad de la entrada (ej: RGB -> 3 canales). En capas intermedias, los canales representan diferentes detectores de características. |
| **Aplanamiento (Flatten)** | Convertir el tensor 2D/3D en un vector 1D para conectarlo a capas densas. |
| **Capa Totalmente Conectada (FC)** | Capa clásica de red neuronal usada al final de la CNN para realizar clasificación o regresión. |

## Arquitectura Típica

1. **Capa de Entrada** (imagen: alto x ancho x canales)
2. **Capa Convolucional + Función de activación** (habitualmente ReLU)
3. **Capa de Pooling** (reduce dimensionalidad)
4. *(Opcional: repetir bloques de Conv+Pool varias veces)*
5. **Aplanamiento (Flatten)**
6. **Capas Densas (FC)**
7. **Capa de Salida** (softmax para clasificación)



## Resumen de Ventajas

- **Parámetros compartidos**: el mismo kernel se aplica en toda la imagen.
- **Sparsity de conexiones**: cada neurona en el feature map solo se conecta a una región local de la entrada.
- **Jerarquía de características**: capas tempranas detectan bordes; capas profundas detectan partes complejas de objetos.
- **Invarianza a traslaciones**: gracias al pooling y al desplazamiento del kernel.

## Fórmula de convolución discreta en 2D

$$
S(x,y) = (w * f)(x, y) = \sum_{i=-a}^{a} \sum_{j=-b}^{b} w(i,j)f(x - i, y - j)
$$

- \( S \): Matriz de salida (Mapa de caracteristicas)
- \( f \): Matriz de entrada (imagen original)
- \( w \): Filtro
- \( (x, y) \): Coordenadas del pixel en el mapa de salida


## Ejemplo de uso en código (Keras/TensorFlow)

```python
# Clasificación de imagenes de numeros

# 1. Importación de librerías necesarias
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from keras import layers, models, regularizers
from keras.utils import to_categorical

# 2. Cargar y preprocesar datos MNIST
(X_train, y_train), (X_val, y_val) = tf.keras.datasets.mnist.load_data()

# Normalización: pasar valores de [0,255] a [0,1] y añadir canal de color (28,28,1)
X_train = X_train.reshape(-1, 28, 28, 1).astype('float32') / 255.0
X_val = X_val.reshape(-1, 28, 28, 1).astype('float32') / 255.0

# Convertir etiquetas a one-hot encoding (necesario para categorical_crossentropy)
y_train = to_categorical(y_train, 10)
y_val = to_categorical(y_val, 10)

print(f"Shape de entrenamiento: {X_train.shape}")  # (60000, 28, 28, 1)
print(f"Shape de validación: {X_val.shape}")        # (10000, 28, 28, 1)

# 3. Visualizar ejemplos del dataset
plt.figure(figsize=(10, 4))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(X_train[i].reshape(28,28), cmap='gray')
    plt.title(f"Clase: {np.argmax(y_train[i])}")
    plt.axis('off')
plt.suptitle("Ejemplos de dígitos manuscritos (MNIST)")
plt.tight_layout()
plt.show()

# 4. Construcción del modelo CNN
model = models.Sequential([
    # Primera capa convolucional + pooling
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),  # Reduce a 14x14
    
    # Segunda capa convolucional + pooling
    layers.Conv2D(64, (3,3), activation='relu'), # 64 filtros de 3x3, Capas de caracteristicas
    layers.MaxPooling2D((2,2)),  # Reduce a 7x7
    
    # Aplanar y capas densas
    layers.Flatten(), # 2D a 1D, con un total de 3136 neuronas
    layers.Dense(64, activation='relu', kernel_initializer='he_normal', kernel_regularizer=regularizers.l2(0.001)),
    layers.Dense(10, activation='softmax')  # 10 clases (dígitos 0-9)
])

# 5. Compilar el modelo
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# 6. Resumen de la arquitectura
model.summary()

# 7. Entrenamiento
history = model.fit(X_train, y_train, 
                    epochs=5, 
                    batch_size=256, 
                    validation_data=(X_val, y_val),
                    verbose=1)

# 8. Evaluación final
val_loss, val_accuracy = model.evaluate(X_val, y_val, verbose=0)
print(f"\nResultados en validación:")
print(f"Pérdida (Loss): {val_loss:.4f}")
print(f"Precisión (Accuracy): {val_accuracy:.4f} ({val_accuracy*100:.2f}%)")

# 9. Visualización de curvas de aprendizaje
plt.figure(figsize=(12, 4))

# Curva de pérdida
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Entrenamiento')
plt.plot(history.history['val_loss'], label='Validación')
plt.title('Pérdida durante el entrenamiento')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

# Curva de precisión
plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Entrenamiento')
plt.plot(history.history['val_accuracy'], label='Validación')
plt.title('Precisión durante el entrenamiento')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# 10. Predicciones y ejemplos de clasificación
predicciones = model.predict(X_val[:10])
clases_predichas = np.argmax(predicciones, axis=1)
clases_reales = np.argmax(y_val[:10], axis=1)

plt.figure(figsize=(12, 4))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(X_val[i].reshape(28,28), cmap='gray')
    color = 'green' if clases_predichas[i] == clases_reales[i] else 'red'
    plt.title(f"Pred: {clases_predichas[i]} | Real: {clases_reales[i]}", color=color)
    plt.axis('off')
plt.suptitle("Predicciones del modelo (verde = correcto, rojo = error)")
plt.tight_layout()
plt.show()
```
